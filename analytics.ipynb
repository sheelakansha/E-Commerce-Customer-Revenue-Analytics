# E-Commerce Customer Revenue Analytics — Visualizations & Insights
This notebook loads the analytical results from PostgreSQL and generates visual charts for revenue trends, cohort retention, and RFM customer segmentation.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load query results
df_monthly = pd.read_csv('results/01_monthly_revenue_growth.csv')
df_cohort = pd.read_csv('results/02_cohort_retention.csv')
df_rfm = pd.read_csv('results/03_rfm_customer_segmentation.csv')
```

## 1. Monthly Revenue & Growth Trend

```python
df_monthly['month'] = pd.to_datetime(df_monthly['month'])
fig, ax1 = plt.subplots(figsize=(10, 5))
ax1.plot(df_monthly['month'], df_monthly['revenue'], color='#1f77b4', marker='o', linewidth=2.5, label='Revenue ($)')
ax1.set_ylabel('Revenue ($)', color='#1f77b4', fontsize=12)
ax2 = ax1.twinx()
ax2.plot(df_monthly['month'], df_monthly['mom_growth_pct'], color='#ff7f0e', linestyle='--', marker='s', label='MoM Growth (%)')
ax2.set_ylabel('MoM Growth (%)', color='#ff7f0e', fontsize=12)
plt.title('Monthly Revenue Trend & MoM Growth Rate', fontsize=14)
plt.show()
```

## 2. Cohort Retention Heatmap

```python
pivot_c = df_cohort.pivot(index='cohort_month', columns='month_number', values='retention_pct')
plt.figure(figsize=(11, 6))
sns.heatmap(pivot_c, annot=True, fmt='.1f', cmap='YlGnBu', vmin=0, vmax=100)
plt.title('Customer Cohort Retention Rate (%)', fontsize=14)
plt.show()
```

## 3. RFM Revenue Distribution

```python
tier_rev = df_rfm.groupby('customer_tier')['monetary'].sum().reset_index()
tier_rev['share_pct'] = (100.0 * tier_rev['monetary'] / tier_rev['monetary'].sum()).round(2)
plt.figure(figsize=(9, 5))
plt.barh(tier_rev['customer_tier'], tier_rev['share_pct'], color='#2ca02c')
plt.xlabel('Revenue Share (%)')
plt.title('Revenue Share by RFM Tier')
plt.show()
```
